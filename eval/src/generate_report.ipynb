```python
import sys
import json
import sqlite3
import pandas as pd

# Reads issues from the local SQLite database (data/review.db).
# The legacy Cosmos DB reader is kept in legacy/ for reference.
```

```python
# Path to the local SQLite database produced by the app (DATA_DIR/review.db).
DB_PATH = "../../data/review.db"


def get_issues_df_from_db(db_path: str) -> pd.DataFrame:
    """Load all issue payloads (JSON documents) into a DataFrame."""
    conn = sqlite3.connect(db_path)
    try:
        rows = conn.execute("SELECT payload FROM issues").fetchall()
    finally:
        conn.close()
    return pd.DataFrame.from_dict([json.loads(r[0]) for r in rows])
```
